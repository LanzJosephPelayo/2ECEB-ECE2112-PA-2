# A. REPRODUCIBLE NORMALIZATION PROBLEM

Code:

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
print("Required Check #A.1:\n","Random 5X5 Array Integer:\n", X)
```

```python
X_normalized=(X-(X.mean()))/(np.std(X))
print("Required Check #A.2:\n","Normalized 5X5 Array Integer Based from the Random Integer:\n",X_normalized)
```

```python
print("Required Check #A.3:\n","Mean of Normalized 5X5 Array Integer:", X_normalized.mean())
```

```python
print("Required Check #A.4:\n","Standard Deviation of Random 5X5 Array Integer:", np.std(X_normalized))
```

Saving the Array:

```python
np.save("X_normalized.npy",X_normalized)
```

#  B.CUBES DIVISIBLE BY 4 PROBLEM

Code:

```python
C = np.arange(1,101,1)
C =  C**3
C = C.reshape(10,10)
print("First 100 Positive Integers raised by 3:\n",C)
print("Required Check #B.1:\n","Shape of Array C: \n", C.shape)
```

```python
div_by_4 = C[C % 4 == 0]
print("Required Check #B.2:\n","The array of Array C that is Divisible by 4:\n", div_by_4)
```

```python
print("Required Check #B.3:\n","The Size of the array of Array C that is Divisible by 4:", div_by_4.size)
```

Saving the Array:

```python
np.save("div_by_4.npy",div_by_4)
```

#  C.ABOVE-MEAN SQUARES PROBLEM

Code:

```python
S = (np.arange(1,37,1)**2).reshape(6,6)
print("Required Check #C.1:\n","The 6x6 Array of First 36 Squared Positive Integers:\n", S)
```

```python
S_mean = S.mean()
print("Required Check #C.2:\n","The Mean of 6x6 Array of The First 36 Squared Positive Integers:\n", S_mean)
```

```python
above_mean = S[S > S_mean]
print("Required Check #C.3:\n","Values of the First 36 Squared Positive Integers that is Greater than the Mean:\n", above_mean)
```

```python
print("Required Check #C.4:\n","The number of Selected Elements/Squared Numbers greater than the Mean:", above_mean.size)
```

Saving the values that is above the mean:

```python
np.save("above_mean.npy",above_mean)
```
